#1.Importing Required Libraries

```python
import pandas as pd

url = "https://raw.githubusercontent.com/datasciencedojo/datasets/master/titanic.csv"

df = pd.read_csv(url)

print("Titanic dataset loaded successfully!")
```

```python
print(df.shape)
```

#4.	Loading the Titanic Dataset

```python
print(df.columns)
```

#5.	Understanding Dataset Information

```python
df.info()
```

#6.Identifying Missing Values

```python
print(df.isnull().sum())
```

#7.Handling missing Age values

```python
df['Age'].fillna(df['Age'].median(), inplace=True)

print("Missing Age values handled!")
```

#8.Handling missing Embarked values

```python
df['Embarked'].fillna(df['Embarked'].mode()[0], inplace=True)

print("Missing Embarked values handled!")
```

#9.Removing Cabin column

```python
df.drop('Cabin', axis=1, inplace=True)

print("Cabin column removed!")
```

# 10. Converting Data Types

```python
print(df.isnull().sum())
```

#11.Identifying Duplicate Records

```python
print("Duplicate rows:", df.duplicated().sum())
```

```python
df.drop_duplicates(inplace=True)

print("Duplicates removed successfully!")
```

#12.Checking and Converting Data Types

```python
df.dtypes
```

#13.Conversion of datatypes

```python
df['Sex'] = df['Sex'].astype('category')

df['Embarked'] = df['Embarked'].astype('category')

print("Data types converted successfully!")
```

```python
df.rename(columns={
    'PassengerId': 'Passenger_ID',
    'Pclass': 'Passenger_Class',
    'SibSp': 'Siblings_Spouses',
    'Parch': 'Parents_Children'
}, inplace=True)

print("Columns renamed successfully!")
```

# 14. Final Dataset Shape

```python
print("Final Dataset Shape:", df.shape)

print("\nMissing Values:")
print(df.isnull().sum())

print("\nDuplicate Rows:")
print(df.duplicated().sum())

df.head()
```

#15.Save the Cleaned Dataset

```python
df.to_csv('Titanic_Cleaned.csv', index=False)

print("Cleaned dataset saved successfully!")
```
